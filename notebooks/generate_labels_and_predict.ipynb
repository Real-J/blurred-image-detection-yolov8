```python
#########

import torch
import os
import glob
from PIL import Image

# Model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

folder_path = '/content/drive/MyDrive/data750/resized'  # Replace with the actual path to the folder

# Get a list of image paths in the folder
image_paths = glob.glob(os.path.join(folder_path, '*.jpg'))  # Modify the extension if necessary

# Load and process the images
imgs = [Image.open(img_path) for img_path in image_paths]

# Inference
results = model(imgs)

# Results
results.print()

# Save annotated images to a separate folder
results_path = '/Users/mathew/Desktop/data750/results'  # Modify this path as needed
os.makedirs(results_path, exist_ok=True)  # Ensure the directory exists
results.save(results_path)  # Save results images

# Save annotations to separate .txt files in a separate folder
annotations_path = '/Users/mathew/Desktop/data750/annotations'  # Modify this path as needed
os.makedirs(annotations_path, exist_ok=True)  # Ensure the directory exists

# Save results to separate .txt files
for i, img_path in enumerate(image_paths):
    basename = os.path.basename(img_path)
    name, ext = os.path.splitext(basename)
    txt_path = os.path.join(annotations_path, name + '.txt')  # Corresponding .txt file path
    with open(txt_path, 'w') as f:
        for *xyxy, conf, _, cls in results.xyxy[i]:  # xyxy are the bounding box coordinates
            x_center = (xyxy[2] + xyxy[0]) / 2  # Calculate x_center
            y_center = (xyxy[3] + xyxy[1]) / 2  # Calculate y_center
            width = xyxy[2] - xyxy[0]  # Calculate width
            height = xyxy[3] - xyxy[1]  # Calculate height
            # Write to the .txt file
            f.write(f"0 {x_center} {y_center} {width} {height}\n")
```
